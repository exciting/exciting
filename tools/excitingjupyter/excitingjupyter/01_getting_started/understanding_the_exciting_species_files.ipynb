### <span style="color:#15317E">1. What is a Species file?</span>

**`exciting`**, as a code based on plane-wave augmentation, distinguishes between the interstitial and muffin-tin regions. In the interstitial region, various quantities, such as wavefunctions, potential, and density are described by means of plane waves. The quality (completeness) of this basis is determined by the maximum wavevector, defined via either <code><span style="color:mediumblue">gmaxvr</span></code> or <code><span style="color:mediumblue">rgkmax</span></code>.  Basis functions in the muffin-tin region are much more cumbersome to define and, thus, require a detailed description. Furthermore, each muffin tin is associated with an atom with a specific nuclear charge, mass, etc. All this information is summarized in species files.

### <span style="color:#15317E">2. Content</span>

Consider a standard species file for carbon.

```xml
<?xml version="1.0" encoding="UTF-8"?>
<spdb xsi:noNamespaceSchemaLocation="../../xml/species.xsd" 
      xmlns:xsi="http://www.w3.org/2001/XMLSchema-instance">
  <sp chemicalSymbol="C" name="carbon" z="-6.00000" mass="21894.16673">
    <muffinTin rmin="0.100000E-04" radius="1.4500" rinf="21.0932" radialmeshPoints="250"/>
    <atomicState n="1" l="0" kappa="1" occ="2.00000" core="true"/>
    <atomicState n="2" l="0" kappa="1" occ="2.00000" core="false"/>
    <atomicState n="2" l="1" kappa="1" occ="1.00000" core="false"/>
    <atomicState n="2" l="1" kappa="2" occ="1.00000" core="false"/>
    <basis>
      <default type="lapw" trialEnergy="0.1500" searchE="false"/>
      <custom l="0" type="apw+lo" trialEnergy="0.1500" searchE="true"/>
      <custom l="1" type="apw+lo" trialEnergy="0.1500" searchE="true"/>
    </basis>
  </sp>
</spdb>
```

The information contained in the first two lines is not relevant for the physics of the system. They are only used to give advanced **XML** instructions to define the **XML** language, style sheet, and schema. For further details on this topic, see [**XML and exciting data format**](https://en.wikipedia.org/wiki/XML). Considering only the physical background of this file and dividing it into segments by adding irrelevant blank lines, the file shown above can be simplified as follows.

```xml
<spdb>
 
   <sp chemicalSymbol="C" name="carbon" z="-6.00000" mass="21894.16673">
 
      <muffinTin rmin="0.100000E-04" radius="1.4500" rinf="21.0932" radialmeshPoints="250"/>
 
      <atomicState n="1" l="0" kappa="1" occ="2.00000" core="true"/>
      <atomicState n="2" l="0" kappa="1" occ="2.00000" core="false"/>
      <atomicState n="2" l="1" kappa="1" occ="1.00000" core="false"/>
      <atomicState n="2" l="1" kappa="2" occ="1.00000" core="false"/>
 
      <basis>
         <default type="lapw" trialEnergy="0.1500" searchE="false"/>
         <custom l="0" type="apw+lo" trialEnergy="0.1500" searchE="true"/>
         <custom l="1" type="apw+lo" trialEnergy="0.1500" searchE="true"/>
      </basis>
 
   </sp>
 
</spdb>
```
This way, we can identify four logical segments.

- The first segment describes the properties of the nucleus.
```xml
  <sp chemicalSymbol="C" name="carbon" z="-6.00000" mass="21894.16673">
```
- The second segment defines the radial grid associated with the atom of this kind. Here <code><span style="color:mediumblue">rmin</span></code> stands for the radial coordinate of the innermost grid point. <code><span style="color:mediumblue">radius</span></code> represents the default value of the muffin-tin radius. Keep in mind that it may be, and frequently is, overridden in the **input.xml** file. <code><span style="color:mediumblue">rinf</span></code> is the coordinate of the effective infinity. This parameter is used in calculations of atoms during the initialization and every time core states are calculated during the self-consistency steps. Finally, <code><span style="color:mediumblue">radialmeshPoints</span></code> defines the number of grid points in the muffin tin. In practice, there will be more points between the muffin-tin boundary and the effective infinity, but their number is calculated automatically. Even though these attributes are not further discussed here, you should be aware that the default choice is not necessarily optimal. The choice of the muffin-tin radius clearly depends on inter-atomic distances and is system-dependent. Additionally, <code><span style="color:mediumblue">rmin</span></code> and <code><span style="color:mediumblue">radialmeshPoints</span></code> are parameters that you might want to explore occasionally.
  
 
```xml
    <muffinTin rmin="0.100000E-04" radius="1.4500" rinf="21.0932" radialmeshPoints="250"/>
```


- The third segment introduces division into core and valence states. The names of attribute are self-explanatory. As the only exception, the meaning of the quantum number <code><span style="color:mediumblue">kappa</span></code> in **`exciting`** differs from that in the textbooks of quantum mechanics where it is defined as $\kappa = (l - j)(2j + 1)$. Here, <code><span style="color:mediumblue">kappa</span></code> adopts the absolute value of the "textbook $\kappa$". According to the standard species file, the shell $1s^2$ is assigned to the core, while $2s^22p^2$ are considered as the valence shells.

```xml
    <atomicState n="1" l="0" kappa="1" occ="2.00000" core="true"/>
    <atomicState n="2" l="0" kappa="1" occ="2.00000" core="false"/>
    <atomicState n="2" l="1" kappa="1" occ="1.00000" core="false"/>
    <atomicState n="2" l="1" kappa="2" occ="1.00000" core="false"/>
```

- The fourth segment describes the muffin-tin basis. The muffin-tin basis can consist of three different kinds of basis functions (**Augmented Plane Waves**, **Linearized Augmented Plane Waves**, and **local orbitals**) which will be explained in detail later in this tutorial. The example below specifies that **apw+lo** is used for the $l=0$ and $l=1$ channels. This is done by adding  <code><span style="color:blue">custom</span></code> elements for these states. All other $l$-channels (up to an maximum $l$) are considered using **lapw** as defined in the <code><span style="color:blue">default</span></code> element. The attribute <code><span style="color:mediumblue">trialEnergy</span></code> defines initial values of the energy parameters which are kept constant during all self-consistent steps. It is possible to adapt them during the calculations by setting the parameter <code><span style="color:mediumblue">searchE</span></code> to **true**. However, this is not recommended since it can lead to unstable calculations. The energy parameters are always considered on the absolute scale, and not as the distance to the Fermi energy.

```xml
    <basis>
      <default type="lapw" trialEnergy="0.1500" searchE="false"/>
      <custom l="0" type="apw+lo" trialEnergy="0.1500" searchE="true"/>
      <custom l="1" type="apw+lo" trialEnergy="0.1500" searchE="true"/>
    </basis>
```

### <span style="color:#15317E">3. LAPW vs. APW+lo</span>

#### <span style="color:#15317E">LAPW</span>

The muffin-tin part of the augmented-plane-wave (**APW**) basis is

\begin{equation}
\tag{1}
\phi_{\textbf{G+k}}(\textbf{r})=\sum_{lm}A^{\textbf{G+k}}_{lm} \; u_l(r;\varepsilon_l) \; Y_{lm}(\hat{\textbf{r}}) \; .
\end{equation}

In order to make **APW** basis functions accurate, the energy parameter $ε_l$ must be a very good approximation to the eigenenergies of states with angular momentum $l$. It is difficult or even impossible to arrange this in practice. By setting a fixed energy parameter $ε_l$, an error called the linearization error is introduced. By adding higher-order energy derivatives $\dot{u_l}=\partial{u_l}/\partial{\epsilon}$ as a Taylor-like expansion into the basis function, the linearization error can be reduced. 

This linearization technique can be implemented using linearized augmented plane waves (**LAPW**). According to **LAPW**, the basis functions in muffin tins are defined as

\begin{equation}
\tag{3}
\phi_{\textbf{G+k}}(\textbf{r})=\sum_{lm}\left[
A^{\textbf{G+k}}_{lm} \; u_l(r;\varepsilon_l) + B^{\textbf{G+k}}_{lm} \; \dot{u_l}(r;\varepsilon_l)
\right] \; Y_{lm}(\hat{\textbf{r}}) \;,
\end{equation}

where the matching coefficients $A^{\textbf{G+k}}_{lm}$ and $B^{\textbf{G+k}}_{lm}$ are chosen in such a way that $\phi_{\textbf{G+k}}(\textbf{r})$ is smooth and continuous at the muffin-tin boundary. To define the smooth (**LAPW**) augmentation in species, set the attribute <code><span style="color:mediumblue">type</span></code> to **"lapw"**. In the example below, the species file for carbon is transformed in such a way that only the augmentation type **LAPW** is used.

```xml
<basis>
      <default type="lapw" trialEnergy="0.15" searchE="false"/>
      <custom l="0" type="lapw" trialEnergy="-0.20" searchE="false"/>
      <custom l="1" type="lapw" trialEnergy="0.15" searchE="false"/>
</basis>
```

Here, the energy parameters used for **LAPW** are $\varepsilon_0=−0.20$, $\varepsilon_1=0.15$ and $\varepsilon_{l>1}=0.15$.


<a id='3.2'></a>
#### <span style="color:#15317E">APW+lo</span>

The alternative approach to reduce the linearization error, is to introduce a different kind of basis function - the local orbital. Local orbitals are defined as

\begin{equation}
\tag{4}
\phi_{\nu}(\textbf{r})=\left[
a_{\nu} \; u_l(r;\varepsilon_l) + b_{\nu} \; \dot{u_l}(r;\varepsilon_l)
\right] \; Y_{lm}(\hat{\textbf{r}}) \;,
\end{equation} 

in one particular muffin-tin and are equal to 0 everywhere else. The example below shows how to transform your basis into an **APW+lo** basis.

```xml
<basis>
      <default type="apw" trialEnergy="0.1500" searchE="false"/>
      <custom l="0" type="apw+lo" trialEnergy="-0.20" searchE="false"/>
      <custom l="1" type="apw+lo" trialEnergy="0.15" searchE="false"/>
</basis>
```

This specification of the basis involves **APW**s for all $l$-channels. However, local orbitals are added for $l=0$ and $l=1$ channels. They are requested by specifying <code><span style="color:mediumblue">type</span></code> = **"apw+lo"** as <code><span style="color:mediumblue">type</span></code> in the custom elements of these states. These local orbitals have exactly the same form as introduced above, namely, they are linear combinations of the "primitive" functions $u_l(r;\varepsilon_l)$ and $\dot{u_l}(r;\varepsilon_l)$. The energy parameters are the same as in the **LAPW** case. In theory **LOs** can also consist of more than two "primitive" functions which can be of higher-order derivatives. These are used to describe semi-core states which are explained in the section on semi-core states. 



<a id='3.3'></a>
#### <span style="color:#15317E">The Best of the two worlds</span>

As you have seen in the example of the first section, **APW+lo** and **LAPW** can be combined.

```xml
<basis>
      <default type="lapw" trialEnergy="0.1500" searchE="false"/>
      <custom l="0" type="apw+lo" trialEnergy="-0.20" searchE="false"/>
      <custom l="1" type="apw+lo" trialEnergy="0.15" searchE="false"/>
</basis>
```

Here, **APW+lo** is used for $l=0$ and $l=1$ channels. Higher $l$ are considered using **LAPW**. 

In general, both **APW+lo** and **LAPW** significantly improve upon **APW**. While, in practice, **APW+lo** is more accurate than **LAPW**, the former methods introduces additional basis functions. Therefore, **APW+lo** is not practical for replacing **APW** for high-$l$ channels. For this reason, we recommend to use **APW+lo** for low $l$, and **LAPW** for high $l$. This strategy is implemented in the standard species files.

### <span style="color:#15317E">4. Semi-core States</span>

The linearization procedure described above is useful for improving the description of valence states. However, local orbitals can be used also for describing semi-core states.

Consider the lithium atom. It has the shell structure $1s^22s^1$, which contains just one valence electron. The $1s$ state can be attributed to the core, but it is neither sufficiently localized within Li muffin tins nor well separated from valence states in terms of energy. Therefore, it makes sense to consider the $1s$ state on the same footing as valence states. This idea is implemented in a species file for Li which you can find below.

```xml
<spdb>
 
  <sp chemicalSymbol="Li" name="lithium" z="-3.00000" mass="12652.66897">
 
    <muffinTin rmin="0.100000E-04" radius="1.7000" rinf="29.9495" radialmeshPoints="250"/>
 
    <atomicState n="1" l="0" kappa="1" occ="2.00000" core="false"/>
    <atomicState n="2" l="0" kappa="1" occ="1.00000" core="false"/>
 
    <basis>
      <default type="lapw" trialEnergy="0.1500" searchE="false"/>
      <custom l="0" type="apw+lo" trialEnergy="0.1500" searchE="true"/>
      <lo l="0">
        <wf matchingOrder="0" trialEnergy="0.1500" searchE="true"/>
        <wf matchingOrder="1" trialEnergy="0.1500" searchE="true"/>
        <wf matchingOrder="0" trialEnergy="-1.8784" searchE="true"/>
      </lo>
    </basis>
 
  </sp>
 
</spdb>
```
The attribute <code><span style="color:mediumblue">core</span></code> = **"false"** explicitly defines that $1s$ and $2s$ are supposed to be treated as valence states. Another important ingredient is the element that describes a local orbital.

```xml
      <lo l="0">
        <wf matchingOrder="0" trialEnergy="0.1500" searchE="true"/>
        <wf matchingOrder="1" trialEnergy="0.1500" searchE="true"/>
        <wf matchingOrder="0" trialEnergy="-1.8784" searchE="true"/>
      </lo>
```

this element consists of three subelements that define three "primitive" functions that satisfy the radial Schrödinger equation. The first subelement contains the attributes <code><span style="color:mediumblue">matchingOrder</span></code> = **"0"** and <code><span style="color:mediumblue">trialEnergy</span></code> = **"0.1500"**. That means that the first "primitive" function corresponds to the zeroth-order energy-derivative and the energy parameter is $0.15$. In other words, the first subelement corresponds to $u_0(r;0.15)$. The second subelement contains <code><span style="color:mediumblue">matchingOrder</span></code> = **"1"**. It means that the second "primitive" function is the first energy-derivative $\dot{u_0}(r;0.15)$. Finally, the third function is $u_0(r;−1.8784)$. Note here the much lower linearization energy that describes the lower-lying semicore state instead of the valence state. 

A more general form of a local orbital describing a semicore state looks like this: 

\begin{equation}
\tag{5}
\phi_{\nu}(\textbf{r})=\left[
a_{\nu} \; u_l(r;\varepsilon^{\text{v}}_l) + b_{\nu} \; \dot{u_l}(r;\varepsilon^{\text{v}}_l) + c_{\nu} \; u_l(r;\varepsilon^{\text{sc}}_{l})
\right] \; Y_{lm}(\hat{\textbf{r}}) \;,
\end{equation} 

where $l$, $\varepsilon^{\text{v}}_l$ is the linearization energy corresponding to the valence state and $\varepsilon^{\text{sc}}_{l}$ is the linearization energy of the semicore state. The matching coefficients $a_{\nu}$, $b_{\nu}$, and $c_{\nu}$ are automatically chosen in a such way that the local orbital normalizes to 1 and turns to 0 at the muffin-tin boundary. These two conditions are essential and are employed for all local orbitals. In order to be able to determine $c_{\nu}$ uniquely, a third condition is applied that forces the radial derivative of the local orbital to turn to 0 at the muffin-tin boundary. It is also possible to construct local orbtials containing other combinations of "primitive" functions with different orders of energy derivatives and linearization energies.

### <span style="color:#15317E">5. Linearization Energies</span>

Choosing appropriate linearization energies for local orbitals can be challenging. To avoid this, linearization energies can be computed automatically. This is done by using the Wigner-Seitz rules. They provide an energy range in which the ideal linearization energy can be found.To get the upper and lower bound energy, the radial Schrödinger equation is solved for the corresponding principal quantum number ($n$), angular momentum ($l$), and initial potential for different energies.The upper bound energy is then equal to the energy for which the radial function is zero at the muffin tin boundary.

\begin{equation}
\tag{6}
u_l(r; E_{\text{max}})\bigg | _{r=R_{\text{MT}}} = 0 \; ,
\end{equation}

The lower bound energy is given by the energy for which the first radial derivative of the radial function vanishes at the muffin tin boundary. 

\begin{equation}
\tag{7}
\frac{\partial u_{l}(r, E_{\text{min}})}{\partial r} \bigg | _{r=R_{\text{MT}}} = 0 \; ,
\end{equation}

In the end, the mean of these two energies is taken as the lineratization energy $\epsilon_{l} = (E_{\text{max}}+E_{\text{min}})/2$.

To utilize this algorithm, the species file must specify the principal quantum number rather than the linearization energy. For the example given above, the <code><span style="color:mediumblue">trialEnergy</span></code> = **"0.1500"** corresponds to the $2s$ state and therefore needs to be replaced by <code><span style="color:mediumblue">n</span></code> = **"2"**. The lower energy <code><span style="color:mediumblue">trialEnergy</span></code> = **"-1.8784"** corresponds to the $1s$ state and needs to be replaced by <code><span style="color:mediumblue">n</span></code> = **"1"**. The whole local orbital subelement then looks like the following.

```xml
      <lo l="0">
        <wf matchingOrder="0" n="2" searchE="false"/>
        <wf matchingOrder="1" n="2" searchE="false"/>
        <wf matchingOrder="0" n="1" searchE="false"/>
      </lo>
```

The calculated linearization energies can be checked in the **LINENGY.OUT** file.

### <span style="color:#15317E">6. Dirac-type Local Orbitals</span>

The standard local orbitals in **`exciting`** are scalar relativistic (**sr**) basis functions. While this is a good approximation for most materials, those with strong spin-orbit coupling (**SOC**) effects are not properly described using a **sr** basis. Therefore, we introduce so called Dirac-type local orbitals.
The "primitive" functions of these Dirac-type local orbitals are calculated using the radial Dirac equation instead of the Schrödinger equation.  Thus, they depend on the total angular momentum $j$.  We use the relativistic quantum number $\kappa$ to define them.

Dirac-type orbitals can be included in the basis by adding a <code><span style="color:green">lo</span></code> subelement with the <code><span style="color:mediumblue">kappa</span></code> attribute to the <code><span style="color:green">basis</span></code> subelement. If we want to include relativistic effects of our $p$-orbital, we can replace our **sr** local orbital 

```xml
      <lo l="1">
        <wf matchingOrder="0" trialEnergy="-4.96" searchE="false"/>
        <wf matchingOrder="1" trialEnergy="-4.96" searchE="false"/>
      </lo>
```

by two local orbitals with the relativistic quantum numbers $\kappa = 1$, corresponding to the $p_{1/2}$ state, and $\kappa = -2$, corresponding to the $p_{3/2}$ state.

```xml
      <lo l="1">
        <wf kappa="1" matchingOrder="0" trialEnergy="-4.96" searchE="false"/>
        <wf kappa="1" matchingOrder="1" trialEnergy="-4.96" searchE="false"/>
      </lo>

      <lo l="1">
        <wf kappa="-2" matchingOrder="0" trialEnergy="-4.56" searchE="false"/>
        <wf kappa="-2" matchingOrder="1" trialEnergy="-4.56" searchE="false"/>
      </lo>
```

However, this is not recommended, as the $p_{3/2}$-orbital (as most other orbitals) is very similar to the **sr** $p$-orbital.
The $p_{1/2}$-orbital, on the other hand, is important because it differs strongly from the **sr** $p$-orbital. Unlike any **sr** orbital, it has a non-zero value at $r=0$ and, therefore, cannot be represented in an **sr** basis. That's why we recommend to add only the $p_{1/2}$-orbitals to an existing **sr** basis set. 
Dirac type local orbitals in general should **always** be used in combination with **second variation with local orbitals**, which can be enabled by setting the input parameter <code><span style="color:mediumblue">svlo</span></code> to **true**. Also the automatic linearization energies explained in Section 5 can not be used for Dirac-type local orbitals. Here, linearization energies must be optimized "by hand". A good starting point for these are the atomic energies written in the **atoms.xml** file.
