### <span style="color:#15317E">1. Produce Coordinate Files of the Animated Phonon Modes</span>

In order to create coordinate files for a given phonon, you should invoke the script **`excitingscripts.plot.phonon_anim`**. This script will read phonon eigenvectors for all q-points contained in the file **PHONON.OUT** and plot a supercell of (**<span style="color:firebrick">default</span>**) **6×6×6** unit cells to visualize them. Unit cells are taken as specified in **input.xml** and the dimensions are along lattice vectors.

````python
%%bash -l
cd run_diamond_phonons
python3 -m excitingscripts.plot.phonon_anim
cd ..
````

If you want to change the supercell size, *e.g.*, type

```bash
python3 -m excitingscripts.plot.phonon_anim -s 4 4 1
```

to obtain a supercell of **4×4×1** unit cells.

<hr style="border:1px solid #DDD"> </hr>

<details>
<summary><strong><span style="color:firebrick">$\Rightarrow$ Further options for <code>excitingscripts.plot.phonon_anim</code></span></strong>
    
</summary> 

The option <code>-n</code> specifies the number of steps in the animation which complete one period of the vibration (**<span style="color:firebrick">default</span>: 20**), whereas the option <code>-c</code> can be used to change the scaling factor (**<span style="color:firebrick">default</span>: 10.0**)
    

</details>
<hr style="border:1px solid #DDD"> </hr>  

The script produces a number of files, named, e.g., **q1_mode1.axsf** and **q1_mode1.xyz**. One file is written for each mode at each q-point. These files contain atomic coordinates for the supercell at every step of the animation. The files with extension axsf are animated **XCrySDen** structure files, the files with extension **xyz** are simple xyz-coordinate files.

The next subsection deals with visualizing these files. To exemplify, we show this using **VMD**. You might have other/additional choices, according to the actual software installed in your environment.
