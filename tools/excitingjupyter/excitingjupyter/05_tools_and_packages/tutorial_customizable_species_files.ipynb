**<span style="color:firebrick">The Purpose of Elemental Crystal Species Files</span>**: In *exciting*, calculations require basis sets&mdash;including local orbitals (LOs)&mdash;tailored to each element.
The files provided here were initially generated for **elemental crystals** to establish a "maximized," highly converged reference basis. 

Depending on your specific target precision, many local orbitals can be pruned to save time. This includes high-energy local orbitals that have negligible impact on ground-state properties, as well as standard local orbitals that can be safely removed if a lower, more lightweight calculation precision is sufficient for your application. 

**The goal of this optimization tool** is to cross-reference these maximized species files with pre-calculated benchmark calculations to dynamically prune non-essential orbitals based on their total energy precision penalty, while strictly protecting those flagged in the benchmark results "Essential" whitelist to preserve the core physics. This yields a mathematically efficient, lightweight basis set that retains the required accuracy and reduces computational time.

*Additional information regarding the species files in general can be found in the tutorial* [understanding_the_exciting_species_files.ipynb](../01_getting_started/understanding_the_exciting_species_files.ipynb)

# Optimizing the Basis Set

As we can see, a lot of local orbitals are included here. On the one hand, we could use this highly converged basis to perform accurate calculations of physical properties. On the other hand, these calculations would be computationally demanding and could also result in diagonalization errors due to numerical linear dependency.

To enhance the performance of our calculation and balance the gap between computational effort and accuracy, we take a look at the pre-generated **local orbital hierarchy**, which can be found in the directory: [lo_hierarchies](../../../excitingscripts/data/lo_hierarchies/)

# Understanding the LO Hierarchy Output

This file contains a wealth of benchmark data. To effectively customize our species files, we focus on **three key metrics**:

---

### 1. DBSV Test Precision
> **Example:** `DBSV test precision for maximized species file: 6.291e-05`

This value provides **insights into the overall quality and completeness** of the basis set currently in use. 
* *For a deeper understanding of the physics, check the **Basis-set completeness** section in the paper:* [An exciting approach to theoretical spectroscopy (arXiv:2601.11388)](https://arxiv.org/pdf/2601.11388)

---

### 2. Orbital Precision
> **Example:** `Ag   2   4,4   1,2   4.097e-04`

This is the **primary key argument** used to dynamically prune the species file. 
* **What it tells us:** It quantifies how much the total energy of the system changes after removing the specific local orbital labeled by its quantum numbers `(l, ns, mOs)`.
* **How to use it:** The baseline *DBSV test precision* serves as a reference point for your threshold. Any local orbital with an energy **above** your chosen threshold is highly significant and **must remain part of the calculation** to preserve accuracy.

---

### 3. Essential Local Orbitals
> **Example:** `Essential local orbitals: ...`

<b> Strictly Protected:</b> These local orbitals are absolutely essential for accurately describing the fundamental chemical properties of the specific element. They form a hard whitelist and <b>cannot be removed</b> under any circumstances.

---

# How to Remove Local Orbitals

To finally remove local orbitals that have low precision values, we use the Python script `optimized_species_files.py`. This script, being executable in the terminal, provides a command-line interface.

The basic syntax is as follows:

```bash
python3 optimized_species_files.py [SPECIES...] [OPTIONS]
````

### Available Arguments & Options
- **Species**
	- **Type:** String
	- **Description:** One or more species names to optimize (e.g. O, Ag, Rh)

- **-p, --precision**
	- **Type:** Float
	- **Default Value:** 1.0e-4
	- **Description:** Precision threshold for removing LOs

- **--path-lo**
	- **Type:** String
	- **Default Value:** `./lo_hierarchies/`
	- **Description:** Root path where the script looks for the `lo_hierarchy_<SPECIES>.out`

- **--path-xml**
	- **Type:** String
	- **Default Value:** `./max_species_files/`
	- **Description:** Root path where the script looks for the maximized species file `<SPECIES>.xml`

- **-o, --output**
	- **Type:** String
	- **Default Value:** `./customized_species/`
	- **Description:** Target directory path where the optimized output files `<SPECIES>_customized` will be written
 


# Example: How to Customize the Species File of Silver

If we want to prune the species file by the default precision threshold of `1.0e-4`, we have to execute the following line in the terminal:

```bash
python3 optimized_species_files.py Ag
```

This gives us the following output: 

```txt

Processing species: Ag
  [DELETED] l=0 ns=(4, 4) mOs=(2, 3) (Precision: 1.00e-08)
  [DELETED] l=0 ns=(4, 4) mOs=(1, 2) (Precision: 4.00e-08)
  [DELETED] l=2 ns=(4, 4) mOs=(2, 3) (Precision: 2.18e-06)
  [DELETED] l=0 ns=(5, 5) mOs=(1, 2) (Precision: 2.72e-05)
----------------------------------------
  [SUCCESS] 4 LOs removed. 

```
We have now successfully deleted four local orbitals that fall below the default precision threshold. This optimization will enhance the computational effort vs. accuracy trade-off. The new species file is saved in the default output path or via the CLI adjustable.
