### <span style="color:#15317E">1. Preparing Input</span>
As part of your tutorial, you will probably want to provide an input file. Make sure the contents of this input file are placed in a separate Markdown cell, with a tag with the following structure placed above it:
```
<span class="TAG_EXAMPLE"></span>
```
To view how this is set up, double-click the cell below that contains the input.

<span class="INPUT_TEMPLATE"></span>
```xml
<input>

   <title>Example input</title>

   <structure speciespath="$EXCITINGROOT/species">

      <crystal scale="6.7274">
         <basevect>0.0   0.5   0.5</basevect>
         <basevect>0.5   0.0   0.5</basevect>
         <basevect>0.5   0.5   0.0</basevect>
      </crystal>

      <species speciesfile="C.xml">
         <atom coord="0.00 0.00 0.00"/>
         <atom coord="0.25 0.25 0.25"/>
      </species>

   </structure>

   <groundstate
      ngridk="4 4 4"
      outputlevel="normal"
      xctype="GGA_PBE_SOL">
   </groundstate>

</input>

```

You can now automatically save these contents to an input file using the `get_input_xml_from_notebook` function. This function takes the tutorial name and the input file tag as arguments, and can be used as shown below.

````python
from excitingjupyter.utilities import get_input_xml_from_notebook

input_str = get_input_xml_from_notebook("tutorial_template", "INPUT_TEMPLATE")

# Write out the input as an XML file:
with open('./run_tutorial_template/input.xml', "w") as fid:
    fid.write(input_str)
````

### <span style="color:#15317E">4. Adding Images</span>
There are several ways to add images to a Jupyter notebook, but in our case, it's important to choose a method that ensures proper rendering when the notebook is converted to **HTML**. To achieve this, we convert all images to Base64 format. You can do this using an online converter, such as: https://www.base64-image.de/

After converting the image to Base64, add the following code to the markdown cell

```
<figure>
<img src="" width="600" align="left"/>
</figure>
```
and add the Base64 code to the `src` attribute of the `<img>` tag.

For more details, check out the following webpage: https://stackoverflow.com/questions/51975699/embedding-image-into-jupyter-notebook-and-exporting-to-html

### <span style="color:#15317E">5. Cell Tags</span>

When converting the notebook to **HTML**, keep in mind that we don’t want to include cells containing Python code, like the one used for saving the input, in the standalone, command-line-based version intended for the website.

To exclude a cell during **HTML** conversion, go to **View → Cell Toolbar → Tags** in the notebook toolbar, then add the tag `remove_cell` to the cell you want to hide. (This has already been done for the relevant cell here.)

Similarly, if you want to hide specific cell outputs from the website version, use the `remove_output` tag. This is already applied to the cell used to run **`exciting`**.

Once you're done adding the tags, return to the normal view by selecting **View → Cell Toolbar → None**.

Once you've completed your tutorial, it can be converted to **HTML** and added to the exciting webpage by the responsible developer. For more details on the **HTML** conversion process, refer to:

